```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# CSV 파일 로드
csv_path = "results/train_speaker_count_eer_multi_dataset.csv"
df = pd.read_csv(csv_path)
df
```

```python
# eval 데이터셋 이름 자동 추출
eval_names = [c.replace('baseline_eer_', '') for c in df.columns if c.startswith('baseline_eer_')]
print(f"Eval datasets: {eval_names}")
```

```python
BASELINE_COLOR = "#9AA0A6"
OURS_COLOR = "#4285F4"

n_datasets = len(eval_names)
fig, axes = plt.subplots(1, n_datasets, figsize=(5 * n_datasets, 5))

if n_datasets == 1:
    axes = [axes]

for ax, name in zip(axes, eval_names):
    bl = df[f'baseline_eer_{name}']
    ours = df[f'ours_eer_{name}']
    ns = df['n_train_speakers']

    ax.set_title(name, fontsize=12, fontweight='bold')

    ax.grid(True, alpha=0.3, linestyle='--')
    ax.plot(ns, bl, color=BASELINE_COLOR, marker='o', markersize=6, linewidth=2, label='Baseline')
    ax.plot(ns, ours, color=OURS_COLOR, marker='s', markersize=6, linewidth=2, label='Ours')
    ax.set_xlabel('# Training Speakers', fontsize=11)
    ax.set_ylabel('EER (%)', fontsize=11)
    ax.legend(loc='best', fontsize=9)
    ax.set_xticks(ns)

    y_min = min(bl.min(), ours.min())
    y_max = max(bl.max(), ours.max())
    y_pad = (y_max - y_min) * 0.15
    ax.set_ylim(max(0, y_min - y_pad), y_max + y_pad)

plt.tight_layout()
save_path = "results/train_speaker_count_eer_multi_dataset.pdf"
plt.savefig(save_path, dpi=300, bbox_inches='tight')
print(f"Saved to: {save_path}")
plt.show()
```
